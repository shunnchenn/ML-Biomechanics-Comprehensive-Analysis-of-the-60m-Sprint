# 3 — What faster looks like

Three interpretable readouts on one feature set, rather than a ranking of
algorithms. With n around 30 the useful output is an effect per feature, not a
leaderboard.

1. **Elastic Net** — LOO-CV R-squared, coefficients on standardised features, so
   each reads as "per +1 SD of this feature, that much change in speed".
2. **Penalised logistic** — fast vs slow outer tertiles, coefficients as log-odds
   per SD.
3. **SHAP** — computed on the linear model, where it is exact. Tree SHAP at this
   sample size would be noise dressed as attribution.

The curve figure uses cluster-based permutation testing (SPM) to mark *where in
the phase* fast and slow athletes actually differ.

Equivalent to `python -m sprint analyse`.

## Coefficients

`beta` is the Elastic Net effect per +1 SD; `lo`/`hi` are percentile bootstrap CIs; `selected_pct` is how often the feature survived the penalty across bootstrap resamples — the most honest stability measure at this sample size.
